```python
import numpy as np
from neuralnet.activation import Sigmoid, ReLu, Softmax
from neuralnet.utils import normalize, one_hot_encoding
from neuralnet.model import Model, Layer
```

```python
def cross_entropy(logits,one_hot_label):
    label = np.where(one_hot_label == 1)[0]
    prod = np.log(logits[np.arange(len(logits)),label])
    return -np.sum(prod)
```

```python
sigmoid_obj = Sigmoid()
sigmoid_obj.forward(np.array([-1,2,3]))
```

```python
relu_obj = ReLu()
post = relu_obj.forward(np.array([-1,2,3]))
relu_obj.backward(post)
```

```python
data = [1,2,134,20,29341,23]
x = np.array(normalize(data))
if x.ndim==1:
    x = np.expand_dims(x,axis=0)
layer1 = Layer(in_feature=x.shape[1], out_feature=10,activation=ReLu())
layer2 = Layer(in_feature=10, out_feature=100,activation=ReLu())
layer3 = Layer(in_feature=100, out_feature=3,activation=Softmax())
```

```python
model = Model(layer1, layer2, layer3)
model.compile(cross_entropy)
```

```python
y = np.array([0])
y = one_hot_encoding(y,num_classes=3)
```

```python
print('Weights before forward')
model.get_weights()
```

```python
logits, loss= model.forward(x,y)
```

```python
print('Weights after forward')
model.get_weights()
```

```python
model.backward()
model.step(0.1)
```

```python
print('Weights after 1 step')
model.get_weights()
```
